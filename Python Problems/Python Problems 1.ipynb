**3-Write a Python program to add 'ing' at the end of a given string. If the given string already ends with 'ing', add 'ly' instead. If the string length is less than 3, leave it unchanged.**

```python
string = input("please enter any string: ")
if len(string) < 3: 
    print(string)
elif string[-3: ] == "ing": 
    print(string + "ly")
else:
    print(string + "ing")
```

**4-Write a Python function that takes a list of words and return the longest word and the length of the longest one**

```python
words = [ "Ahmed", "Ali" , "Mohamed", "Ibrahimovic" , "Essam" , "Emad"]
```

```python
def return_word ( words_list : list) :
    max_word = words_list[0]
    for word in words_list:
        if len(word) > len(max_word): 
            max_word = word
    return max_word, len(max_word)
```

```python
return_word(words)
```
