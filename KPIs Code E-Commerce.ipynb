<center> <h1> E-Commerce KPIs </h1> </center>

```python
import pandas as pd
data = pd.read_csv("ecommerce_orders.csv")
data.head()
```

1. Total Revenue

```python
total_revenue = data["total_amount"].sum()
print(total_revenue)
```

2. Conversion Rate

- Formula : Purchases / Sessions


```python
conversion_rate = (
    data["purchased"].sum()
    /
    data["session_id"].nunique()
) * 100

print(conversion_rate)
```

3. Average Order Value (AOV)

- Formula : Revenue / Number of Orders

```python
aov = (
    data["total_amount"].sum()
    /
    data["order_id"].nunique()
)

print(aov)
```

4. Customer Lifetime Value (CLV)

- Simple version : Revenue per Customer

```python
customer_revenue = (
    data.groupby("customer_id")["total_amount"]
      .sum()
)

clv = customer_revenue.mean()

print(clv)
```

5. Customer Acquisition Cost (CAC)

- Formula : Marketing Cost / New Customers

```python
marketing_cost = data["marketing_cost"].sum()

new_customers = data["customer_id"].nunique()

cac = marketing_cost / new_customers

print(cac)
```

6. Cart Abandonment Rate

- Formula : (Carts - Purchases) / Carts    

```python
carts = data["cart_created"].sum()

completed = data["cart_purchased"].sum()

cart_abandonment = (
    (carts - completed)
    /
    carts
) * 100

print(cart_abandonment)
```

7. Repeat Purchase Rate

- Formula : Customers with > 1 Order / Total Customers

```python
orders_per_customer = (
    data.groupby("customer_id")
      .size()
)

repeat_customers = (
    orders_per_customer > 1
).sum()

total_customers = data["customer_id"].nunique()

repeat_rate = (
    repeat_customers
    /
    total_customers
) * 100

print(repeat_rate)
```

8. Gross Merchandise Value (GMV)

- Formula : Sum of all sold products

```python
gmv = (
    data["quantity"]
    *
    data["unit_price"]
).sum()

print(gmv)
```

9. Refund Rate

- Formula : Refunded Orders / Total Orders

```python
refund_rate = (
    data["refunded"].sum()
    /
    data["order_id"].nunique()
) * 100

print(refund_rate)
```

10. Net Profit Margin

- Formula : (Net Profit / Revenue)

- Net Profit : Revenue - Cost

```python
revenue = data["total_amount"].sum()

cost = data["cost"].sum()

profit = revenue - cost

net_profit_margin = (
    profit
    /
    revenue
) * 100

print(net_profit_margin)
```
