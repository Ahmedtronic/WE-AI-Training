<center> <h1>KPIs In Marketing </h1> </center>

```python
import pandas as pd 
df = pd.read_csv("marketing_data.csv") 
df 
```

1. Marketing ROI: 

- Formula : ((Revenue - Ad Spend) / Ad Spend) × 100     

```python
marketing_roi = (
    (df["revenue"].sum() - df["ad_spend"].sum())
    /
    df["ad_spend"].sum()
) * 100

print(marketing_roi)
```

2. ROAS
   
- Formula : Revenue / Ad Spend

```python
roas = (
    df["revenue"].sum()
    /
    df["ad_spend"].sum()
)

print(roas)
```

3. Cost Per Lead (CPL)
- Formula : Ad Spend / Leads 

```python
cpl = (
    df["ad_spend"].sum()
    /
    df["leads"].sum()
)

print(cpl)
```

4. Cost Per Acquisition (CPA)
- Formula : Ad Spend / Customers

```python
cpa = (
    df["ad_spend"].sum()
    /
    df["customers"].sum()
)

print(cpa)
```

5. Click Through Rate (CTR)
- Formula : Clicks / Impressions

```python
ctr = (
    df["clicks"].sum()
    /
    df["impressions"].sum()
) * 100

print(ctr)
```

6. Lead Conversion Rate
- Formula : Customers / Leads

```python
lead_conversion = (
    df["customers"].sum()
    /
    df["leads"].sum()
) * 100

print(lead_conversion)
```

7. Customer Retention Rate
- Formula : Retained Customers / Total Customers

```python
retention_rate = (
    df["retained_customers"].sum()
    /
    df["total_customers"].sum()
) * 100

print(retention_rate)
```

8. Brand Awareness Score
- Assume surveys already produce a score from 0–100.

```python
brand_awareness = (
    df["brand_score"].mean()
)

print(brand_awareness)
```

9. Share of Voice
- Formula: Brand Mentions / Industry Mentions

```python
share_of_voice = (
    df["brand_mentions"].sum()
    /
    df["industry_mentions"].sum()
) * 100

print(share_of_voice)
```

10. Engagement Rate
Formula : (Likes + Comments + Shares) / Followers

```python
engagement_rate = (
    (
        df["likes"].sum()
        +
        df["comments"].sum()
        +
        df["shares"].sum()
    )
    /
    df["followers"].sum()
) * 100

print(engagement_rate)
```
