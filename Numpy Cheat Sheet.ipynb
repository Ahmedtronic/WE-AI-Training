# NumPy Cheat Sheet

## Import
```python
import numpy as np
```

## Array Creation
```python
np.array([1, 2, 3])                 # from list
np.zeros((3, 4))                    # 3x4 array of zeros
np.ones((2, 3))                     # 2x3 array of ones
np.full((2, 2), 7)                  # filled with 7
np.eye(3)                           # 3x3 identity matrix
np.arange(0, 10, 2)                 # [0, 2, 4, 6, 8]
np.linspace(0, 1, 5)                # 5 evenly spaced values 0-1
np.random.rand(2, 3)                # uniform [0,1)
np.random.randn(2, 3)               # standard normal
np.random.randint(0, 10, (2, 3))    # random ints
np.empty((2, 2))                    # uninitialized
```

## Array Attributes
```python
a.shape        # dimensions, e.g. (3, 4)
a.ndim         # number of dimensions
a.size         # total number of elements
a.dtype        # data type
a.itemsize     # bytes per element
a.nbytes       # total bytes
```

## Reshaping
```python
a.reshape(3, 4)          # new shape (same data)
a.reshape(-1, 1)         # column vector, auto-infer
a.flatten()              # 1D copy
a.ravel()                # 1D view (no copy if possible)
a.T                      # transpose
a.transpose(1, 0, 2)     # custom axis order
np.expand_dims(a, axis=0)
np.squeeze(a)            # remove size-1 dims
```

## Indexing & Slicing
```python
a[0]              # first element/row
a[-1]             # last element/row
a[1:4]            # slice
a[:, 0]           # first column (2D)
a[0, :]           # first row (2D)
a[1:3, 1:3]       # sub-matrix
a[::-1]           # reverse
a[a > 5]           # boolean mask
np.where(a > 5)    # indices where condition true
np.where(a > 5, 1, 0)  # conditional replace
```

## Combining Arrays
```python
np.concatenate([a, b], axis=0)
np.vstack([a, b])        # stack vertically
np.hstack([a, b])        # stack horizontally
np.stack([a, b], axis=0) # new axis stack
np.split(a, 3)           # split into 3 arrays
np.array_split(a, 3)     # split unevenly if needed
```

## Math Operations
```python
a + b, a - b, a * b, a / b     # elementwise
a ** 2                          # power
np.sqrt(a)
np.exp(a)
np.log(a)
np.abs(a)
a @ b or np.dot(a, b)           # matrix multiplication
np.matmul(a, b)
np.round(a, 2)
np.clip(a, 0, 1)                # clamp values
```

## Aggregations
```python
a.sum()               # total sum
a.sum(axis=0)         # column sums
a.sum(axis=1)         # row sums
a.mean()
a.std()
a.var()
a.min(), a.max()
a.argmin(), a.argmax()   # index of min/max
np.median(a)
np.percentile(a, 90)
a.cumsum()             # cumulative sum
a.cumprod()            # cumulative product
```

## Comparison & Logic
```python
np.array_equal(a, b)
np.any(a > 5)
np.all(a > 0)
np.isnan(a)
np.isinf(a)
np.logical_and(a > 1, a < 5)
```

## Linear Algebra (np.linalg)
```python
np.linalg.inv(a)          # inverse
np.linalg.det(a)          # determinant
np.linalg.eig(a)          # eigenvalues/vectors
np.linalg.svd(a)          # singular value decomposition
np.linalg.norm(a)         # vector/matrix norm
np.linalg.solve(A, b)     # solve Ax = b
np.trace(a)               # sum of diagonal
```

## Broadcasting Rules
```python
# Shapes align from the right; dims must match or be 1
a.shape == (3, 4)
b.shape == (4,)      # broadcasts across rows
a + b                 # works: b applied to each row

a.shape == (3, 1)
b.shape == (1, 4)
a + b                 # broadcasts to (3, 4)
```

## Copy vs View
```python
b = a.view()      # shares memory (changes affect both)
b = a.copy()       # independent copy
```

## Useful Utilities
```python
np.unique(a)              # sorted unique values
np.sort(a)                 # sorted copy
a.sort()                   # in-place sort
np.argsort(a)               # indices that would sort
np.repeat(a, 3)             # repeat elements
np.tile(a, (2, 2))          # tile array
np.diff(a)                  # discrete differences
np.cross(a, b)              # cross product
np.outer(a, b)              # outer product
np.meshgrid(x, y)           # coordinate grids
np.set_printoptions(precision=3, suppress=True)
```

## Random (Generator API — modern approach)
```python
rng = np.random.default_rng(seed=42)
rng.random((2, 3))
rng.integers(0, 10, size=5)
rng.normal(loc=0, scale=1, size=10)
rng.choice([1, 2, 3], size=5, replace=True)
rng.shuffle(a)
```

## Saving & Loading
```python
np.save('arr.npy', a)
a = np.load('arr.npy')
np.savetxt('arr.csv', a, delimiter=',')
a = np.loadtxt('arr.csv', delimiter=',')
```

# Common Axis Rule

| Operation | axis=0 | axis=1 |
|-----------|--------|--------|
| Sum | Down columns | Across rows |
| Mean | Per column | Per row |
| Max | Max of each column | Max of each row |
| Min | Min of each column | Min of each row |
